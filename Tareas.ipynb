```python
# Fahrenheit a Celsius

fahrenheit_celsius = lambda f: (float(f) - 32) * 5 / 9

f = float(input("Ingrese los grados Fahrenheit: "))
print("Grados Celsius:", fahrenheit_celsius(f))
```

```python
# Celsius a Kelvin

celsius_kelvin = lambda c: float(c) + 273.15

c = float(input("Ingrese los grados Celsius: "))
print("Grados Kelvin:", celsius_kelvin(c))
```

```python
# Función anónima para calcular el salario bruto

salario = lambda horas, valor: horas * valor * 30

horas = float(input("Horas trabajadas por día: "))
valor = float(input("Valor por hora: "))

bruto = salario(horas, valor)

# Descuentos
isss = bruto * 0.03
afp = bruto * 0.0725

# Base para calcular ISR
base = bruto - isss - afp

# Cálculo del ISR según la tabla mensual
if base <= 550.00:
    isr = 0
elif base <= 895.24:
    isr = ((base - 550.00) * 0.10) + 17.67
elif base <= 2038.10:
    isr = ((base - 895.24) * 0.20) + 60.00
else:
    isr = ((base - 2038.10) * 0.30) + 288.57

neto = base - isr

print("\n------ DESPRENDIBLE DE PAGO ------")
print("Salario Bruto: $", round(bruto, 2))
print("ISSS (3%): $", round(isss, 2))
print("AFP (7.25%): $", round(afp, 2))
print("ISR: $", round(isr, 2))
print("Salario Neto: $", round(neto, 2))
```

```python
# Elevar un número a su exponente

potencia = lambda numero, exponente: float(numero) ** float(exponente)

numero = float(input("Ingrese el número: "))
exponente = float(input("Ingrese el exponente: "))

print("Resultado:", potencia(numero, exponente))
```

**MEDIA ARMONICA**

```python
from functools import reduce

class estadistica:

    def mediaArmonica(self, serie):
        suma = reduce(lambda acum, num: acum + (1 / float(num)), serie, 0)
        return len(serie) / suma


objEstadistica = estadistica()

serie = input("Serie: ").split(",")

print("Media armonica: ", objEstadistica.mediaArmonica(serie))
```

**ENCONTRAR LA FRECUENCIA**

```python
class estadistica:

    def frecuencia(self, serie):
        for num in serie:
            cantidad = serie.count(num)
            print(num, "aparece", cantidad, "veces")


objEstadistica = estadistica()

serie = input("Serie: ").split(",")

objEstadistica.frecuencia(serie)
```

**PALINDROMO**

```python
class palindromo:

    def verificar(self, palabra):
        palabra = palabra.lower()

        if palabra == palabra[::-1]:
            return True
        else:
            return False


objPalindromo = palindromo()

palabra = input("Palabra: ")

if objPalindromo.verificar(palabra):
    print("La palabra es palindromo")
else:
    print("La palabra no es palindromo")
```

```python
import math
from collections import Counter

estudiantes = [
    {"nombre": "Karen Elizabeth Ulloa Gomez", "nota": 8.5, "edad": 20, "traslado hacia la UGB": 45, "Tiempo en el celular": 4.5},
    {"nombre": "Maria", "nota": 9.0, "edad": 19, "traslado": 30, "telefono": 3.0},
    {"nombre": "Luis", "nota": 6.5, "edad": 21, "traslado": 60, "telefono": 5.0},
    {"nombre": "Ana", "nota": 10.0, "edad": 20, "traslado": 20, "telefono": 2.5},
    {"nombre": "Jose", "nota": 7.0, "edad": 22, "traslado": 90, "telefono": 6.0},
    {"nombre": "Elena", "nota": 8.0, "edad": 19, "traslado": 40, "telefono": 4.0},
    {"nombre": "Diego", "nota": 5.5, "edad": 20, "traslado": 50, "telefono": 5.5},
    {"nombre": "Lucia", "nota": 9.5, "edad": 21, "traslado": 15, "telefono": 2.0},
    {"nombre": "Kevin", "nota": 7.5, "edad": 20, "traslado": 35, "telefono": 3.5},
    {"nombre": "Sofia", "nota": 8.5, "edad": 19, "traslado": 25, "telefono": 4.0},
    {"nombre": "Mateo", "nota": 6.0, "edad": 23, "traslado": 70, "telefono": 5.0},
    {"nombre": "Camila", "nota": 9.0, "edad": 20, "traslado": 30, "telefono": 3.0},
    {"nombre": "Gabriel", "nota": 7.0, "edad": 21, "traslado": 45, "telefono": 4.0},
    {"nombre": "Valeria", "nota": 8.0, "edad": 20, "traslado": 60, "telefono": 3.5},
    {"nombre": "Daniel", "nota": 4.5, "edad": 22, "traslado": 80, "telefono": 7.0},
    {"nombre": "Paula", "nota": 9.0, "edad": 19, "traslado": 20, "telefono": 2.0},
    {"nombre": "Andres", "nota": 7.5, "edad": 20, "traslado": 40, "telefono": 4.5},
    {"nombre": "Fernanda", "nota": 8.5, "edad": 21, "traslado": 35, "telefono": 3.0},
    {"nombre": "Jorge", "nota": 6.5, "edad": 20, "traslado": 50, "telefono": 5.0},
    {"nombre": "Andrea", "nota": 9.5, "edad": 19, "traslado": 15, "telefono": 1.5}
]


def media(lista):
    return sum(lista) / len(lista)


def mediana(lista):
    ordenados = sorted(lista)
    n = len(lista)
    mitad = n // 2
    if n % 2 == 0:
        return (ordenados[mitad - 1] + ordenados[mitad]) / 2
    else:
        return ordenados[mitad]


def moda(lista):
    conteo = Counter(lista)
    max_veces = max(conteo.values())
    if max_veces == 1:
        return "No hay moda"
    resultado = []
    for num, veces in conteo.items():
        if veces == max_veces:
            resultado.append(str(num) + " (se repite " + str(veces) + " veces)")
    return resultado


def varianza(lista):
    m = media(lista)
    suma = 0
    for x in lista:
        suma += (x - m) ** 2
    return suma / (len(lista) - 1)


def desviacion_estandar(lista):
    return math.sqrt(varianza(lista))


def desviacion_tipica(lista):
    return math.sqrt(varianza(lista))


def rango(lista):
    return max(lista) - min(lista)


def frecuencia(lista):
    return dict(Counter(lista))


def mostrar_nombres_y_valores(campo):
    print("")
    for est in estudiantes:
        print(est["nombre"], "->", est[campo])


def elegir_campo():
    print("")
    print("De cual dato quieres ver esto:")
    print("1. Notas")
    print("2. Edades")
    print("3. Tiempo de traslado")
    print("4. Horas de telefono")
    op = input("Elige una opcion: ")

    if op == "1":
        return "nota"
    elif op == "2":
        return "edad"
    elif op == "3":
        return "traslado"
    elif op == "4":
        return "telefono"
    else:
        print("Opcion no valida")
        return None


# Menu principal
while True:
    print("")
    print("Sistema estadistico, calcule:")
    print("1. Media aritmetica")
    print("2. Mediana")
    print("3. Moda")
    print("4. Varianza")
    print("5. Desviacion estandar")
    print("6. Desviacion tipica")
    print("7. Rango")
    print("8. Frecuencia")
    print("0. Salir")

    opcion = input("Elige una opcion: ")

    if opcion == "0":
        break

    campo = elegir_campo()
    if campo is None:
        continue

    mostrar_nombres_y_valores(campo)

    valores = [est[campo] for est in estudiantes]

    print("")
    if opcion == "1":
        print("Media:", round(media(valores), 2))
    elif opcion == "2":
        print("Mediana:", round(mediana(valores), 2))
    elif opcion == "3":
        print("Moda:", moda(valores))
    elif opcion == "4":
        print("Varianza:", round(varianza(valores), 2))
    elif opcion == "5":
        print("Desviacion estandar:", round(desviacion_estandar(valores), 2))
    elif opcion == "6":
        print("Desviacion tipica:", round(desviacion_tipica(valores), 2))
    elif opcion == "7":
        print("Rango:", rango(valores))
    elif opcion == "8":
        print("Frecuencia:", frecuencia(valores))
    else:
        print("Opcion no valida")
```

```python
import pandas as pd
```

```python
archivo = "/content/notas.csv"
notas = pd.read_csv(archivo, sep=";", encoding="utf-8")
print("Hola ", notas["nombre"][0], "te sacaste",notas["nota"][0],"en el lab1.")
```

```python
aprobados = notas[notas["nota"] >= 6]
print(aprobados)
```

```python
reprobados = notas[notas["nota"] < 6]
print(reprobados)
```

```python
print( "cantidad de alumnos aprobados " , len(aprobados))
print( "cantidad de alumnos reprobados " , len(reprobados))
```

```python
cantidad_aprobados = notas.loc[notas["nota"] >= 6, "nota"].count()
cantidad_reprobados = notas.loc[notas["nota"] < 6, "nota"].count()

print("Cantidad de alumnos aprobados:", cantidad_aprobados)
print("Cantidad de alumnos reprobados:", cantidad_reprobados)
```

```python
cantidad_aprobados = notas.loc[notas["nota"] >= 6].agg({"nota": "count"})
cantidad_reprobados = notas.loc[notas["nota"] < 6].agg({"nota": "count"})

print("Cantidad de alumnos aprobados:", cantidad_aprobados)
print("Cantidad de alumnos reprobados:", cantidad_reprobados)
```

```python
notas_alumnos = notas.groupby(["nota"]).agg({'nombre': lambda n: ', '.join(map(str, n))})
print(notas_alumnos)
```

```python
#importar la libreria para graficos
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
notas['estado'] = notas['nota'].apply(lambda x: 'Aprobado' if x >= 6 else 'Reprobado')
notas
```

```python
sns.set(style="whitegrid")
sns.countplot(
    x="estado",
    data=notas,
    hue="estado",
    palette={'Aprobado': 'green', 'Reprobado': 'red'}
    )
plt.title("Distribución de Notas")
plt.xlabel("Estado")
plt.ylabel("Cantidad")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Contar la cantidad de alumnos por estado
conteo_estados = notas['estado'].value_counts()

# Configurar el gráfico de pastel
plt.figure(figsize=(6, 6))
plt.pie(
    conteo_estados,
    labels=conteo_estados.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=['green', 'red']
)
plt.title('Porcentaje de Alumnos Aprobados vs. Reprobados')
plt.show()
```

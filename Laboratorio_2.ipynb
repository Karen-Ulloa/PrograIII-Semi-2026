**1.** Calcular sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y sueldo neto. (20%)

```python
print("CALCULO DE DEDUCCIONES")

sueldo = float(input("Ingrese el sueldo: $"))

isss = sueldo * 0.03
afp = sueldo * 0.0625

if sueldo <= 472:
    isr = 0
elif sueldo <= 895.24:
    isr = ((sueldo - 472) * 0.10) + 17.67
elif sueldo <= 2038.10:
    isr = ((sueldo - 895.24) * 0.20) + 60
else:
    isr = ((sueldo - 2038.10) * 0.30) + 288.57

total_deducciones = isss + afp + isr
sueldo_neto = sueldo - total_deducciones

print("\nRESULTADOS")
print("ISSS: $", round(isss, 2))
print("AFP: $", round(afp, 2))
print("ISR: $", round(isr, 2))
print("Total de deducciones: $", round(total_deducciones, 2))
print("Sueldo neto: $", round(sueldo_neto, 2))
```

**2.** Calcular el valor a pagar según el número de metros consumidos de agua potable (Ver tabla en la siguiente diapositiva). (20%)

```python
print("CALCULO DE AGUA POTABLE")

metros = float(input("Ingrese los metros consumidos: "))

if metros <= 18:
    total = 6

elif metros <= 28:
    exceso = metros - 18
    total = 6 + (exceso * 0.45)

else:
    exceso1 = 28 - 18
    exceso2 = metros - 28

    total = 6 + (exceso1 * 0.45) + (exceso2 * 0.65)

print("Valor a pagar: $", round(total, 2))
```

**5.** Realizar el sistema de gestión del tiempo y la demanda en las unidades de salud.

```python
pacientes = []
seguir = "s"

while seguir == "s":

    print("\n===================================")
    print("         SISTEMA DE TRIAGE")
    print("===================================")

    nombre = input("Nombre del paciente: ")
    edad = int(input("Edad: "))

    # Determinar tipo de paciente
    if edad < 1:
        tipo = "Bebé"
    elif edad < 12:
        tipo = "Niño"
    elif edad < 65:
        tipo = "Adulto"
    else:
        tipo = "Adulto mayor"

    print("Tipo de paciente:", tipo)

    # Motivo de consulta
    print("\n--- MOTIVO DE CONSULTA ---")
    print("1. Problema respiratorio")
    print("2. Dolor")
    print("3. Fiebre")
    print("4. Accidente o lesión")
    print("5. Problema digestivo")
    print("6. Embarazo")
    print("7. Otro")

    motivo = input("Seleccione una opción: ")

    riesgo = 3
    orden = 1
    motivo_nombre = "No especificado"

    # ==========================================
    # PROBLEMA RESPIRATORIO
    # ==========================================

    if motivo == "1":

        motivo_nombre = "Problema respiratorio"

        print("\n--- PROBLEMA RESPIRATORIO ---")

        dificultad = input(
            "¿Tiene dificultad para respirar? s/n: "
        ).lower()

        labios = input(
            "¿Presenta labios morados o azulados? s/n: "
        ).lower()

        no_puede_hablar = input(
            "¿Tiene dificultad para hablar por falta de aire? s/n: "
        ).lower()

        if labios == "s" and no_puede_hablar == "s":
            riesgo = 1
            orden = 4

        elif labios == "s" or no_puede_hablar == "s":
            riesgo = 1
            orden = 3

        elif dificultad == "s":
            riesgo = 2
            orden = 3

        else:
            riesgo = 3
            orden = 1


    # ==========================================
    # DOLOR
    # ==========================================

    elif motivo == "2":

        motivo_nombre = "Dolor"

        print("\n--- PROBLEMA DE DOLOR ---")

        lugar = input("¿Dónde presenta el dolor?: ")

        intensidad = int(
            input("Del 1 al 10, ¿qué tan fuerte es el dolor?: ")
        )

        pecho = input(
            "¿El dolor es en el pecho? s/n: "
        ).lower()

        perdida_conocimiento = input(
            "¿Ha perdido el conocimiento? s/n: "
        ).lower()

        if perdida_conocimiento == "s":
            riesgo = 1
            orden = 4

        elif pecho == "s" and intensidad >= 7:
            riesgo = 1
            orden = 4

        elif pecho == "s":
            riesgo = 1
            orden = 3

        elif intensidad >= 8:
            riesgo = 2
            orden = 3

        elif intensidad >= 4:
            riesgo = 2
            orden = 2

        else:
            riesgo = 3
            orden = 1


    # ==========================================
    # FIEBRE
    # ==========================================

    elif motivo == "3":

        motivo_nombre = "Fiebre"

        print("\n--- PROBLEMA DE FIEBRE ---")

        temperatura = float(
            input("Ingrese la temperatura: ")
        )

        convulsiones = input(
            "¿Ha presentado convulsiones? s/n: "
        ).lower()

        estado = input(
            "¿Está muy decaído o no responde normalmente? s/n: "
        ).lower()

        if convulsiones == "s" and estado == "s":
            riesgo = 1
            orden = 4

        elif convulsiones == "s":
            riesgo = 1
            orden = 4

        elif estado == "s":
            riesgo = 1
            orden = 3

        elif temperatura >= 40:
            riesgo = 1
            orden = 4

        elif temperatura >= 39:
            riesgo = 2
            orden = 3

        elif temperatura >= 38:
            riesgo = 2
            orden = 2

        else:
            riesgo = 3
            orden = 1


    # ==========================================
    # ACCIDENTE O LESIÓN
    # ==========================================

    elif motivo == "4":

        motivo_nombre = "Accidente o lesión"

        print("\n--- ACCIDENTE O LESIÓN ---")

        sangrado = input(
            "¿Tiene sangrado abundante? s/n: "
        ).lower()

        perdida_conocimiento = input(
            "¿Perdió el conocimiento? s/n: "
        ).lower()

        dificultad = input(
            "¿Tiene dificultad para respirar? s/n: "
        ).lower()

        fractura = input(
            "¿Presenta una posible fractura? s/n: "
        ).lower()

        if sangrado == "s" and perdida_conocimiento == "s":
            riesgo = 1
            orden = 4

        elif sangrado == "s" or perdida_conocimiento == "s":
            riesgo = 1
            orden = 4

        elif dificultad == "s":
            riesgo = 1
            orden = 3

        elif fractura == "s":
            riesgo = 2
            orden = 3

        else:
            riesgo = 3
            orden = 1


    # ==========================================
    # PROBLEMA DIGESTIVO
    # ==========================================

    elif motivo == "5":

        motivo_nombre = "Problema digestivo"

        print("\n--- PROBLEMA DIGESTIVO ---")

        sangre = input(
            "¿Tiene sangre en el vómito o en las heces? s/n: "
        ).lower()

        dolor = input(
            "¿Tiene dolor abdominal muy fuerte? s/n: "
        ).lower()

        vomitos = input(
            "¿Tiene vómitos repetitivos? s/n: "
        ).lower()

        if sangre == "s" and dolor == "s":
            riesgo = 1
            orden = 4

        elif sangre == "s":
            riesgo = 1
            orden = 4

        elif dolor == "s":
            riesgo = 1
            orden = 3

        elif vomitos == "s":
            riesgo = 2
            orden = 2

        else:
            riesgo = 3
            orden = 1


    # ==========================================
    # EMBARAZO
    # ==========================================

    elif motivo == "6":

        motivo_nombre = "Embarazo"

        print("\n--- EMBARAZO ---")

        sangrado = input(
            "¿Presenta sangrado? s/n: "
        ).lower()

        dolor = input(
            "¿Tiene dolor fuerte? s/n: "
        ).lower()

        desmayo = input(
            "¿Ha presentado desmayo? s/n: "
        ).lower()

        if sangrado == "s" and desmayo == "s":
            riesgo = 1
            orden = 4

        elif desmayo == "s":
            riesgo = 1
            orden = 4

        elif sangrado == "s":
            riesgo = 1
            orden = 3

        elif dolor == "s":
            riesgo = 2
            orden = 2

        else:
            riesgo = 3
            orden = 1


    # ==========================================
    # OTRO
    # ==========================================

    elif motivo == "7":

        motivo_nombre = "Otro"

        print("\n--- OTRO MOTIVO ---")

        emergencia = input(
            "¿Presenta una emergencia? s/n: "
        ).lower()

        sintomas_importantes = input(
            "¿Presenta síntomas importantes? s/n: "
        ).lower()

        if emergencia == "s":
            riesgo = 1
            orden = 3

        elif sintomas_importantes == "s":
            riesgo = 2
            orden = 2

        else:
            riesgo = 3
            orden = 1


    else:

        print("Opción no válida.")
        motivo_nombre = "No especificado"
        riesgo = 3
        orden = 1


    # ==========================================
    # DETERMINAR PRIORIDAD
    # ==========================================

    if riesgo == 1:
        prioridad = "Prioridad máxima"
        tiempo = "Atención inmediata"

    elif riesgo == 2:
        prioridad = "Prioridad intermedia"
        tiempo = "Atención en menos de 2 horas"

    else:
        prioridad = "Prioridad menor"
        tiempo = "Atención en 2 horas, máximo 4 horas"


    # ==========================================
    # GUARDAR PACIENTE
    # ==========================================

    paciente = {
        "nombre": nombre,
        "edad": edad,
        "tipo": tipo,
        "motivo": motivo_nombre,
        "riesgo": riesgo,
        "orden": orden,
        "prioridad": prioridad,
        "tiempo": tiempo
    }

    pacientes.append(paciente)


    # ==========================================
    # ORDENAR PACIENTES
    # ==========================================

    pacientes.sort(
        key=lambda x: (x["riesgo"], -x["orden"])
    )


    # ==========================================
    # RESULTADO
    # ==========================================

    print("\n===================================")
    print("       RESULTADO DEL TRIAGE")
    print("===================================")

    print("Paciente:", nombre)
    print("Edad:", edad)
    print("Tipo:", tipo)
    print("Motivo:", motivo_nombre)
    print("Riesgo:", riesgo)
    print("Prioridad:", prioridad)
    print("Tiempo:", tiempo)


    # ==========================================
    # LISTA DE TRIAGE
    # ==========================================

    print("\n===================================")
    print("          LISTA DE TRIAGE")
    print("===================================")

    for i, paciente in enumerate(pacientes, 1):

        print(
            i, "-",
            paciente["nombre"],
            "| Edad:", paciente["edad"],
            "|", paciente["tipo"],
            "| Motivo:", paciente["motivo"],
            "| Riesgo:", paciente["riesgo"]
        )


    seguir = input(
        "\n¿Desea registrar otro paciente? s/n: "
    ).lower()


# ==========================================
# MODIFICAR RIESGO
# ==========================================

if len(pacientes) > 0:

    modificar = input(
        "\n¿Desea modificar el riesgo de algún paciente? s/n: "
    ).lower()

    while modificar == "s":

        print("\n===================================")
        print("       PACIENTES REGISTRADOS")
        print("===================================")

        for i, paciente in enumerate(pacientes, 1):

            print(
                i, "-",
                paciente["nombre"],
                "| Motivo:", paciente["motivo"],
                "| Riesgo:", paciente["riesgo"]
            )


        numero = int(
            input("\nIngrese el número del paciente: ")
        )


        if numero >= 1 and numero <= len(pacientes):

            nuevo_riesgo = int(
                input("Ingrese el nuevo riesgo (1, 2 o 3): ")
            )


            if nuevo_riesgo == 1:

                pacientes[numero - 1]["riesgo"] = 1
                pacientes[numero - 1]["prioridad"] = "Prioridad máxima"
                pacientes[numero - 1]["tiempo"] = "Atención inmediata"


            elif nuevo_riesgo == 2:

                pacientes[numero - 1]["riesgo"] = 2
                pacientes[numero - 1]["prioridad"] = "Prioridad intermedia"
                pacientes[numero - 1]["tiempo"] = "Atención en menos de 2 horas"


            elif nuevo_riesgo == 3:

                pacientes[numero - 1]["riesgo"] = 3
                pacientes[numero - 1]["prioridad"] = "Prioridad menor"
                pacientes[numero - 1]["tiempo"] = "Atención en 2 horas, máximo 4 horas"


            else:

                print("Riesgo no válido.")


            # Reordenar después de modificar
            pacientes.sort(
                key=lambda x: (x["riesgo"], -x["orden"])
            )


            print("\n--- LISTA ACTUALIZADA ---")

            for i, paciente in enumerate(pacientes, 1):

                print(
                    i, "-",
                    paciente["nombre"],
                    "| Edad:", paciente["edad"],
                    "|", paciente["tipo"],
                    "| Motivo:", paciente["motivo"],
                    "| Riesgo:", paciente["riesgo"]
                )


        else:

            print("Número de paciente no válido.")


        modificar = input(
            "\n¿Desea modificar otro paciente? s/n: "
        ).lower()


# ==========================================
# LISTA FINAL
# ==========================================

print("\n===================================")
print("        LISTA FINAL DE TRIAGE")
print("===================================")

for i, paciente in enumerate(pacientes, 1):

    print(
        i, "-",
        paciente["nombre"],
        "| Edad:", paciente["edad"],
        "|", paciente["tipo"],
        "| Motivo:", paciente["motivo"],
        "| Riesgo:", paciente["riesgo"]
    )

print("\nSistema de triage finalizado.")
```
